```python
from pathlib import Path

# Facebook AI Similarity Search (faiss) is a library for efficient similarity search and clustering of dense vectors. 
import faiss

# faiss uses NumPy as its memory-bridge to Python.
import json, numpy as np 

input_dir = Path(
    "/content/drive/MyDrive/DS RAG-PROJECT/embeddings_data"
)

# creating faiss directory for storing embeddings...
faiss_dir = Path(
    "/content/drive/MyDrive/DS RAG-PROJECT/faiss_index"
)

faiss_dir.mkdir(
    parents=True,
    exist_ok=True
)

all_chunks = []

for file in input_dir.glob("*.json"):
  with open(file, 'r', encoding='utf-8') as f:
    chunks = json.load(f)

  all_chunks.extend(chunks)  # Add each chunk from chunks to the all_chunks list.

# converting embeddings into numpy array([[]])...
vectors = np.array(
    [chunk["embedding"] for chunk in all_chunks],
    dtype="float32"
)

# printing the shape of vector -> (860, 384)...
print(f"Vector shape: {np.shape(vectors)}")

# Get the dimension (number of values) in each vector...
dimension = vectors.shape[1]

# Create a FAISS index that expects vectors of this dimension.
# IndexFlatL2 uses Euclidean (L2) distance to compare vectors.
index = faiss.IndexFlatL2(dimension)

# Add all vectors to the FAISS index for similarity search.
index.add(vectors)

index_path = faiss_dir / "index.faiss"

# Save the FAISS index to disk.
faiss.write_index(
    index,
    str(index_path) # Convert the path to a string because FAISS expects a string path.
)

# Save chunks for rag retrieval in future...
chunks_path = faiss_dir / "chunks.json"

with open(chunks_path, 'w', encoding='utf-8') as f:
  json.dump(all_chunks, f, ensure_ascii=False, indent=4)


```
