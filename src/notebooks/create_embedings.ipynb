```python
from pathlib import Path
import json

# for embeddings..
from sentence_transformers import SentenceTransformer

input_dir = Path(
    "/content/drive/MyDrive/DS RAG-PROJECT/chunks_data"
)
output_dir = Path(
    "/content/drive/MyDrive/DS RAG-PROJECT/embeddings_data"
)

output_dir.mkdir(
    parents=True,
    exist_ok=True
)

# loading embeddings model...
model = SentenceTransformer(
    "sentence-transformers/all-MiniLM-L6-v2"
)


# loop through each file...
for file in input_dir.glob("*.json"):
  embedded_chunks = []

  with open(file, 'r', encoding='utf-8') as f:
    chunks = json.load(f)

  # extracting text...
  texts = [chunk["text"] for chunk in chunks]

  # generating embeddings...
  embeddings = model.encode(
      texts,
      show_progress_bar=True
  )

  # Pair each chunk with its corresponding embedding.
  # zip() makes sure that each chunk gets the embedding generated for it.
  for chunk, embedding in zip(chunks, embeddings):
    embedded_chunks.append({
        "text": chunk["text"],
        "embedding": embedding.tolist(),
        "metadata": chunk["metadata"]
    })

  # specifying files names and path...
  output_path = output_dir / f"{file.stem}_embeddings.json"

  # Save the chunks, embeddings, and metadata to a JSON file.
  # encoding='utf-8' → supports different characters and languages
  # ensure_ascii=False → keeps non-ASCII characters readable
  # indent=4 → formats the JSON with 4-space indentation
  with open(output_path, 'w', encoding='utf-8') as f:
    json.dump(embedded_chunks, f, ensure_ascii=False, indent=4)


  print(
    f"Processed: {file.stem} | "
    f"Chunks: {len(embedded_chunks)}"
)


print("\nAll embeddings generated! 🎉")
```
